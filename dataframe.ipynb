```python
df.isna() #Checks the table for NaN values
```

```python
df.isna().sum() #Sum shows the axis = 0 each column and then add them according to column
```

```python
columns = df.isna().any() #Checks which columns have at least one null value and any is used for True or False (default, axis = 0)
```

Removing Missing Data

```python
df #Works on basis of rows 
```

```python
df.dropna() # Drop is always based on rows (removes rows completely if they have na)
```

```python
df.dropna(thresh=3,axis = 0) #Dropna with the threshold of 3 looks for minimum 3 non null values and the remaining it will remove  
```

```python
df.dropna(thresh=4,axis = 0) #Dropna will know find rows that have at least 4 non null values
```

```python
pd.merge(employees,salaries, on="employee_id", how = "outer")
```

Groupby and Aggregation

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.groupby("Category")["Quantity"].mean()
```

```python
df[df["Category"] == "A"][["Quantity"]].sum() #Took rows of Categories that are A and the quantity column of the  and the sum of them 
```

```python
df[df["Category"] == "A"]["Quantity"]
```

Aggregation

```python
df
```

```python
grouped = df.groupby("Store")[["Quantity","Sales"]].agg(["sum","mean"])
```

```python
grouped
```
